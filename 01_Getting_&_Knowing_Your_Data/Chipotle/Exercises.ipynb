### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 



```python
path = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"  # tsv stands for tab-separated values, so in this case you have to use sep="\t"

df = pd.read_csv(path, sep="\t")
```

```python
chipo = df.copy()
```

### Step 13. Turn the item price into a float



```python
chipo["item_price"] = chipo["item_price"].apply(lambda x: float(x[1:]))
chipo 

```

#### Step 13 Check the item price type

### Step 14. How much was the revenue for the period in the dataset?

```python
chipo["revenue"] = chipo["quantity"] * chipo["item_price"]
tot_revenue = chipo["revenue"].sum()
print(f"the total revenue for the period is ${tot_revenue}")
```

### Step 15. How many orders were made in the period?

```python
tot_orders = chipo["order_id"].nunique()

print(f"{tot_orders} orders have been made in total")
```
